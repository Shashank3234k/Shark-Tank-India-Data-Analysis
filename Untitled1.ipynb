```python
shark_summary = pd.DataFrame({
    "Shark": ["Ashneer","Anupam","Aman","Namita","Vineeta","Peyush","Ghazal"],
    "Deals": [
        df["ashneer_deal"].sum(),
        df["anupam_deal"].sum(),
        df["aman_deal"].sum(),
        df["namita_deal"].sum(),
        df["vineeta_deal"].sum(),
        df["peyush_deal"].sum(),
        df["ghazal_deal"].sum()
    ]
})

shark_summary.sort_values("Deals", ascending=False)
```
